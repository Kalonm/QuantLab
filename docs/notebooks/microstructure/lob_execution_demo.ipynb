# Limit Order Book Simulation

This notebook demonstrates how to generate synthetic order flow and evaluate an execution strategy.

```python
from src.microstructure.lob import LimitOrderBook, OrderFlowSimulator, Side
from src.microstructure.execution import ExecutionAnalytics

simulator = OrderFlowSimulator(limit_rate=6.0, market_rate=2.0, cancel_rate=1.5, price_grid=[99.5, 99.75, 100.0, 100.25, 100.5], seed=42)
book = simulator.simulate(10.0)
analytics = ExecutionAnalytics(book, simulator)

result = analytics.estimate_fill_probability(Side.BUY, quantity=5, horizon=5.0, n_paths=50)
impact = analytics.estimate_price_impact(Side.BUY, quantity=5)

result, impact
```
