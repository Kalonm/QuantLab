# Point Process Estimators

This notebook illustrates how to calibrate Poisson and Hawkes processes using the lightweight estimators provided in `QuantLab`.

```python
import numpy as np
from src.microstructure.point_process import PoissonProcess, HawkesProcess

rng = np.random.default_rng(7)
poisson_events = np.cumsum(rng.exponential(0.5, size=200))
poisson_model = PoissonProcess().fit(poisson_events)

hawkes_events = np.cumsum(rng.exponential(0.35, size=200))
hawkes_model = HawkesProcess().fit(hawkes_events)

print("Estimated Poisson rate:", poisson_model.rate)
print("Estimated Hawkes params:", hawkes_model.params)
```
